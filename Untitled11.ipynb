```python
# HÜCRE 2 — İmportlar
import os, json, math, warnings, random
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image
from collections import Counter
from sklearn.metrics import classification_report, confusion_matrix
warnings.filterwarnings('ignore')

import tensorflow as tf
from tensorflow.keras.applications import EfficientNetV2S
from tensorflow.keras.applications.efficientnet_v2 import preprocess_input
from tensorflow.keras import layers
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping

tf.keras.mixed_precision.set_global_policy('float32')

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
set_seed(42)

print(f'✅ TensorFlow: {tf.__version__}')
print(f'   GPU: {tf.config.list_physical_devices("GPU")}')
print(f'   CPU kullanılıyor')
```

```python
# HÜCRE 4 — Konfigürasyon
VERI        = '/content/drive/MyDrive/tarihi_yapilar2'
KAYIT       = '/content/drive/MyDrive/tarihi_yapi_model_v322'
DURUM_DOSYA = f'{KAYIT}/durum.json'
os.makedirs(KAYIT, exist_ok=True)

TRAIN_DIR = os.path.join(VERI, 'train')
VALID_DIR = os.path.join(VERI, 'valid')
TEST_DIR  = os.path.join(VERI, 'test')

IMG_SIZE       = (224, 224)
BATCH_SIZE     = 8
ASAMA1_EPOCH   = 25
ASAMA2_EPOCH   = 60
MIXUP_ALPHA    = 0.3
PATIENCE_1     = 8
PATIENCE_2     = 20

# Sınıf sayısı — direkt klasörden say
NUM_CLASSES = len([d for d in os.listdir(TRAIN_DIR)
                   if os.path.isdir(os.path.join(TRAIN_DIR, d))])

METRIKLER = [
    'accuracy',
    tf.keras.metrics.TopKCategoricalAccuracy(k=5, name='top5_acc')
]

print(f'✅ Veri yolu    : {VERI}')
print(f'✅ Kayıt yolu   : {KAYIT}')
print(f'✅ Sınıf sayısı : {NUM_CLASSES}')
print(f'✅ Görüntü boyutu: {IMG_SIZE}')
print(f'✅ Batch size   : {BATCH_SIZE}')
```

```python
def ornekleri_say(klasor):
    exts = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'}
    sayim = {}
    for cls in sorted(os.listdir(klasor)):
        yol = os.path.join(klasor, cls)
        if os.path.isdir(yol):
            n = sum(1 for f in os.listdir(yol)
                    if Path(f).suffix.lower() in exts)
            sayim[cls] = n
    return sayim

train_sayim = ornekleri_say(TRAIN_DIR)
valid_sayim = ornekleri_say(VALID_DIR)
test_sayim  = ornekleri_say(TEST_DIR)

print(f'Train : {sum(train_sayim.values())} görüntü')
print(f'Valid : {sum(valid_sayim.values())} görüntü')
print(f'Test  : {sum(test_sayim.values())} görüntü')
print(f'\nTrain — min: {min(train_sayim.values())}, '
      f'max: {max(train_sayim.values())}, '
      f'ort: {np.mean(list(train_sayim.values())):.1f}')

az = {k: v for k, v in train_sayim.items() if v < 5}
if az:
    print(f'\n⚠️  5 altı görüntülü {len(az)} sınıf:')
    for k, v in sorted(az.items(), key=lambda x: x[1]):
        print(f'   {k}: {v}')

# Dağılım grafiği
vals = sorted(train_sayim.values())
plt.figure(figsize=(16, 3))
plt.bar(range(len(vals)), vals, color='steelblue', alpha=0.8)
plt.axhline(5,  color='red',    ls='--', label='Min (5)')
plt.axhline(10, color='orange', ls='--', label='Hedef (10)')
plt.xlabel('Sınıf'); plt.ylabel('Görüntü sayısı')
plt.title('Train Seti Dağılımı'); plt.legend()
plt.tight_layout()
plt.savefig(f'{KAYIT}/dagilim.png', dpi=100)
plt.show()
```

```python
# HÜCRE 6 — MixUp (orijinal, sadece bir kez)
def mixup_batch(images, labels, alpha=MIXUP_ALPHA):
    batch_size = tf.shape(images)[0]
    lam = tf.cast(
        np.random.beta(alpha, alpha), tf.float32
    )
    idx = tf.random.shuffle(tf.range(batch_size))
    mixed_images = lam * images + (1 - lam) * tf.gather(images, idx)
    mixed_labels = lam * labels + (1 - lam) * tf.gather(labels, idx)
    return mixed_images, mixed_labels

print(f'✅ MixUp hazırlandı (alpha={MIXUP_ALPHA})')
```

```python
def tf_augment(image, label):
    """TF native augmentation — albumentations olmadan."""
    image = tf.cast(image, tf.float32)
    image = tf.image.random_flip_left_right(image)
    image = tf.image.random_brightness(image, max_delta=0.25)
    image = tf.image.random_contrast(image, lower=0.75, upper=1.25)
    image = tf.image.random_saturation(image, lower=0.8, upper=1.2)
    image = tf.clip_by_value(image, 0, 255)
    return image, label

print('✅ Augmentation pipeline hazırlandı')
```

```python
# HÜCRE 7 — Dataset Pipeline (orijinal)
print('Veri yükleniyor...')

train_raw = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=True, seed=42
)
valid_raw = tf.keras.utils.image_dataset_from_directory(
    VALID_DIR,
    image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=False
)
test_raw = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=False
)

CLASS_NAMES = train_raw.class_names
np.save(f'{KAYIT}/class_names.npy', CLASS_NAMES)
print(f'✅ Sınıf sayısı: {len(CLASS_NAMES)}')

train_ds = (
    train_raw
    .map(tf_augment, num_parallel_calls=tf.data.AUTOTUNE)
    .map(lambda x, y: (preprocess_input(x), y),
         num_parallel_calls=tf.data.AUTOTUNE)
    .map(lambda x, y: mixup_batch(x, y),
         num_parallel_calls=tf.data.AUTOTUNE)
    .shuffle(512)
    .prefetch(tf.data.AUTOTUNE)
)
valid_ds = (
    valid_raw
    .map(lambda x, y: (preprocess_input(x), y),
         num_parallel_calls=tf.data.AUTOTUNE)
    .prefetch(tf.data.AUTOTUNE)
)
test_ds = (
    test_raw
    .map(lambda x, y: (preprocess_input(x), y),
         num_parallel_calls=tf.data.AUTOTUNE)
    .prefetch(tf.data.AUTOTUNE)
)

print('✅ Dataset pipeline hazırlandı')
```

```python
# HÜCRE 8 — Model
def model_olustur(num_classes, dropout=0.6):
    base = EfficientNetV2S(
        input_shape=(*IMG_SIZE, 3),
        include_top=False,
        weights='imagenet',
        pooling=None
    )
    base.trainable = False

    inputs = tf.keras.Input(shape=(*IMG_SIZE, 3))
    x = base(inputs, training=False)

    gap = layers.GlobalAveragePooling2D()(x)
    gmp = layers.GlobalMaxPooling2D()(x)
    x   = layers.Concatenate()([gap, gmp])

    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout)(x)
    x = layers.Dense(
        512, activation='relu',
        kernel_regularizer=tf.keras.regularizers.l2(1e-3)
    )(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout * 0.5)(x)
    x = layers.Dense(
        256, activation='relu',
        kernel_regularizer=tf.keras.regularizers.l2(1e-3)
    )(x)
    x = layers.Dropout(dropout * 0.25)(x)
    outputs = layers.Dense(num_classes, activation='softmax',
                           dtype='float32')(x)

    return tf.keras.Model(inputs, outputs)

print('✅ Model fonksiyonu hazırlandı')
```

```python
# HÜCRE 9 — WarmupCosineDecay
class WarmupCosineDecay(tf.keras.callbacks.Callback):
    def __init__(self, total_epochs, warmup_epochs, base_lr, min_lr=1e-7):
        super().__init__()
        self.total_epochs  = total_epochs
        self.warmup_epochs = warmup_epochs
        self.base_lr       = base_lr
        self.min_lr        = min_lr

    def _lr(self, epoch):
        if epoch < self.warmup_epochs:
            return self.base_lr * (epoch + 1) / self.warmup_epochs
        prog = (epoch - self.warmup_epochs) / max(
            1, self.total_epochs - self.warmup_epochs)
        return self.min_lr + 0.5 * (self.base_lr - self.min_lr) * (
            1 + math.cos(math.pi * prog))

    def on_epoch_begin(self, epoch, logs=None):
        self.model.optimizer.learning_rate.assign(self._lr(epoch))

    def on_epoch_end(self, epoch, logs=None):
        print(f'  LR: {float(self.model.optimizer.learning_rate):.2e}')

print('✅ WarmupCosineDecay hazırlandı')
```

```python
# HÜCRE 10 — Durum Yönetimi
def durum_oku():
    if os.path.exists(DURUM_DOSYA):
        with open(DURUM_DOSYA) as f:
            return json.load(f)
    return {'asama': 1, 'tamamlanan_epoch': 0}

def durum_kaydet(asama, epoch):
    with open(DURUM_DOSYA, 'w') as f:
        json.dump({'asama': asama, 'tamamlanan_epoch': epoch}, f)

class DurumCallback(tf.keras.callbacks.Callback):
    def __init__(self, asama, baslangic):
        self.asama     = asama
        self.baslangic = baslangic
    def on_epoch_end(self, epoch, logs=None):
        durum_kaydet(self.asama, self.baslangic + epoch + 1)

durum = durum_oku()
print(f'✅ Durum: Aşama {durum["asama"]}, '
      f'tamamlanan epoch: {durum["tamamlanan_epoch"]}')
```

```python
# HÜCRE 11 — Aşama 1
if durum['asama'] == 1:
    print('=' * 60)
    print('AŞAMA 1: Head eğitimi (backbone dondurulmuş)')
    print('=' * 60)

    asama1_yolu = f'{KAYIT}/asama1_model.keras'

    if os.path.exists(asama1_yolu) and durum['tamamlanan_epoch'] > 0:
        print(f'Kaldığı yerden devam '
              f'({durum["tamamlanan_epoch"]}/{ASAMA1_EPOCH})...')
        model = tf.keras.models.load_model(asama1_yolu)
    else:
        print('Yeni model oluşturuluyor...')
        model = model_olustur(NUM_CLASSES)

    kalan     = ASAMA1_EPOCH - durum['tamamlanan_epoch']
    baslangic = durum['tamamlanan_epoch']

    model.compile(
        optimizer=tf.keras.optimizers.Adam(3e-4, clipnorm=1.0),
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=METRIKLER
    )

    print(f'Toplam parametre : {model.count_params():,}')
    print(f'Eğitilecek epoch : {kalan}')

    history1 = model.fit(
        train_ds, validation_data=valid_ds,
        epochs=kalan,
        callbacks=[
            WarmupCosineDecay(total_epochs=kalan, warmup_epochs=3,
                              base_lr=3e-4, min_lr=1e-5),
            ModelCheckpoint(asama1_yolu, monitor='val_accuracy',
                            save_best_only=True, verbose=1),
            EarlyStopping(monitor='val_accuracy', patience=PATIENCE_1,
                          restore_best_weights=True, verbose=1),
            DurumCallback(asama=1, baslangic=baslangic),
        ],
        verbose=1
    )

    best = max(history1.history['val_accuracy'])
    print(f'\n✅ Aşama 1 tamamlandı — val_acc: {best:.4f} ({best*100:.1f}%)')
    durum_kaydet(asama=2, epoch=0)
    durum = durum_oku()
```

```python
# HÜCRE 12 — Aşama 2
if durum['asama'] == 2:
    print('=' * 60)
    print('AŞAMA 2: Fine-Tuning (son 50 katman açıldı)')
    print('=' * 60)

    son_model_yolu   = f'{KAYIT}/son_model.keras'
    final_model_yolu = f'{KAYIT}/final_model.keras'
    asama1_yolu      = f'{KAYIT}/asama1_model.keras'

    if durum['tamamlanan_epoch'] > 0 and os.path.exists(son_model_yolu):
        print(f'Kaldığı yerden devam '
              f'({durum["tamamlanan_epoch"]}/{ASAMA2_EPOCH})...')
        model = tf.keras.models.load_model(son_model_yolu)
    else:
        print('Aşama 1 modeli yükleniyor...')
        model = tf.keras.models.load_model(asama1_yolu)

    base = model.get_layer('efficientnetv2-s')
    base.trainable = True
    for layer in base.layers[:-50]:
        layer.trainable = False
    for layer in base.layers:
        if isinstance(layer, layers.BatchNormalization):
            layer.trainable = False

    acik = sum(1 for l in base.layers if l.trainable)
    print(f'Açık katman: {acik}/{len(base.layers)}')

    kalan     = ASAMA2_EPOCH - durum['tamamlanan_epoch']
    baslangic = durum['tamamlanan_epoch']

    model.compile(
        optimizer=tf.keras.optimizers.Adam(5e-6, clipnorm=1.0),
        loss=tf.keras.losses.CategoricalCrossentropy(),
        metrics=METRIKLER
    )

    history2 = model.fit(
        train_ds, validation_data=valid_ds,
        epochs=kalan,
        callbacks=[
            WarmupCosineDecay(total_epochs=kalan, warmup_epochs=5,
                              base_lr=5e-6, min_lr=1e-8),
            ModelCheckpoint(final_model_yolu, monitor='val_accuracy',
                            save_best_only=True, verbose=1),
            ModelCheckpoint(son_model_yolu,
                            save_best_only=False, verbose=0),
            EarlyStopping(monitor='val_accuracy', patience=PATIENCE_2,
                          restore_best_weights=True, verbose=1),
            DurumCallback(asama=2, baslangic=baslangic),
        ],
        verbose=1
    )

    durum_kaydet(asama=3, epoch=ASAMA2_EPOCH)
    best_acc  = max(history2.history['val_accuracy'])
    best_top5 = max(history2.history['val_top5_acc'])

    print('\n' + '=' * 60)
    print('EĞİTİM TAMAMLANDI')
    print(f'  val_accuracy Top-1 : {best_acc:.4f}  ({best_acc*100:.1f}%)')
    print(f'  val_accuracy Top-5 : {best_top5:.4f}  ({best_top5*100:.1f}%)')
    print(f'  Model              : {final_model_yolu}')
    print('=' * 60)
```

```python
# HÜCRE 13 — Test Değerlendirmesi (eğitim bittikten sonra çalıştırın)
final_model_yolu = f'{KAYIT}/final_model.keras'
model = tf.keras.models.load_model(final_model_yolu)
print(f'✅ Model yüklendi')

results = model.evaluate(test_ds, verbose=1)
for name, val in zip(model.metrics_names, results):
    print(f'  {name}: {val:.4f}  ({val*100:.1f}%)')

all_preds, all_labels = [], []
for images, labels in test_ds:
    preds = model.predict(images, verbose=0)
    all_preds.extend(np.argmax(preds, axis=1))
    all_labels.extend(np.argmax(labels.numpy(), axis=1))

print(classification_report(
    all_labels, all_preds,
    target_names=CLASS_NAMES,
    zero_division=0
))
```

```python
# ONNX'e çevir ve Drive'a kaydet
!pip install tf2onnx --quiet

import tf2onnx
import tensorflow as tf
import numpy as np
import os

KAYIT = '/content/drive/MyDrive/tarihi_yapi_model_v322'

# Modeli yükle
model = tf.keras.models.load_model(f'{KAYIT}/final_model.keras')
print('✅ Model yüklendi')

# ONNX'e çevir
onnx_yolu = f'{KAYIT}/eser_tanima.onnx'

model_proto, _ = tf2onnx.convert.from_keras(
    model,
    input_signature=[tf.TensorSpec(shape=(None, 224, 224, 3),
                                   dtype=tf.float32, name='input')],
    opset=13,
    output_path=onnx_yolu
)

boyut = os.path.getsize(onnx_yolu) / (1024 * 1024)
print(f'✅ ONNX model kaydedildi: {boyut:.1f} MB')
print(f'   Konum: {onnx_yolu}')

# İndirmek istersen
from google.colab import files
files.download(onnx_yolu)
```
